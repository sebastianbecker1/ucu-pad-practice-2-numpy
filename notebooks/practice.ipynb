## Ejercicio 1

```python
import numpy as np

# Fijamos la semilla aleatoria para garantizar la reproducibilidad
np.random.seed(42)

# Definimos el tamaño de la muestra
n_observaciones = 1000

# 1. Generación de la Distribución Normal (Gaussiana)
# loc = media, scale = desvío estándar
datos_normal = np.random.normal(loc=50, scale=10, size=n_observaciones)

# 2. Generación de la Distribución Uniforme
# low = límite inferior, high = límite superior
datos_uniforme = np.random.uniform(low=10, high=90, size=n_observaciones)

# Verificación rápida del tamaño generado
print(f"Tamaño de datos normales: {len(datos_normal)}")
print(f"Tamaño de datos uniformes: {len(datos_uniforme)}")
```

## Ejercicio 2

```python
# 2. Cálculo de estadísticos descriptivos

# Estadísticos para la Distribución Normal
estadisticos_normal = {
    "Media": np.mean(datos_normal),
    "Mediana": np.median(datos_normal),
    "Varianza": np.var(datos_normal),
    "Desvío Estándar": np.std(datos_normal),
    "Mínimo": np.min(datos_normal),
    "Máximo": np.max(datos_normal)
}

# Estadísticos para la Distribución Uniforme
estadisticos_uniforme = {
    "Media": np.mean(datos_uniforme),
    "Mediana": np.median(datos_uniforme),
    "Varianza": np.var(datos_uniforme),
    "Desvío Estándar": np.std(datos_uniforme),
    "Mínimo": np.min(datos_uniforme),
    "Máximo": np.max(datos_uniforme)
}

# Mostrar los resultados en una tabla formateada
print("=" * 52)
print(f"{'ESTADÍSTICO':<20} | {'NORMAL':<12} | {'UNIFORME':<12}")
print("=" * 52)
for clave in estadisticos_normal:
    print(f"{clave:<20} | {estadisticos_normal[clave]:12.4f} | {estadisticos_uniforme[clave]:12.4f}")
print("=" * 52)
```

## Ejercicio 3


```python
import matplotlib.pyplot as plt

# Crear la figura y los subplots (1 fila, 2 columnas)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Subplot Izquierdo: Histograma de la Distribución Normal
axes[0].hist(datos_normal, bins=30, color='skyblue', edgecolor='black', alpha=0.75, label='Normal (μ=50, σ=10)')
axes[0].set_title('Distribución Normal (Gaussiana)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Valor', fontsize=10)
axes[0].set_ylabel('Frecuencia', fontsize=10)
axes[0].legend(loc='upper right')
axes[0].grid(True, linestyle='--', alpha=0.6)

# Subplot Derecho: Histograma de la Distribución Uniforme
axes[1].hist(datos_uniforme, bins=30, color='lightgreen', edgecolor='black', alpha=0.75, label='Uniforme [10, 90]')
axes[1].set_title('Distribución Uniforme', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Valor', fontsize=10)
axes[1].set_ylabel('Frecuencia', fontsize=10)
axes[1].legend(loc='upper right')
axes[1].grid(True, linestyle='--', alpha=0.6)

# Título general de la figura
plt.suptitle('Comparación de Distribuciones Probabilísticas (n = 1.000)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```
